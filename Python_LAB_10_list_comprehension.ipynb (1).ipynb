```python
trains = {
    "Rajdhani": 120,
    "Duronto": 95,
    "Garib Rath": 40,
    "Shatabdi": 60
}

# Dictionary comprehension
available_trains = {
    train: seats
    for train, seats in trains.items()
    if seats > 50
}

# Display result
print("Trains With More Than 50 Seats:")
print(available_trains)
```

```python
def calculate_fare(ticket_price, passengers):
   total_amount = ticket_price * passengers
   return total_amount
```

```python
def ticket_details(**details):
   print("Ticket Information:")
   for key, value in details.items():
       print(key, ":", value)
```

```python
ticket_details(
   passenger_name="Rahul Sharma",
   train_name="Rajdhani Express",
   seat_number="B1-21",
   fare=2500,
   status="Confirmed")
```

```python
def seat_countdown(seats):
   if seats == 0:
       print("All Seats Booked")
       return
   print("Remaining Seats :", seats)
   seat_countdown(seats - 1)
```

```python
seat_countdown(5)
```

```python
def verify_passenger(passenger_number):
   if passenger_number == 0:
       print("Verification Completed")
       return
   print("Verifying Passenger :", passenger_number)
   verify_passenger(passenger_number - 1)
```

```python
verify_passenger(3)
```

```python
def final_booking(
   train_name="Rajdhani Express",
   *seat_numbers
):
   print("===================================")
   print(" Railway Reservation ")
   print("===================================")
   print("Train Name :", train_name)
   for seat in seat_numbers:
       print("Allocated Seat :", seat)
```

```python
final_booking(
   "Rajdhani Express",
   "B1-21",
   "B1-22",
   "B1-23")
```
